```python
# Check geographic coordinate ranges across all monthly datasets
for filename in sorted(os.listdir(csv_path)):
    if filename.endswith(".csv"):
        monthly_df = pd.read_csv(f"{csv_path}/{filename}")

        invalid_start_coords = (
            (monthly_df["start_lat"].abs() > 90) |
            (monthly_df["start_lng"].abs() > 180)
        ).sum()

        invalid_end_coords = (
            (monthly_df["end_lat"].abs() > 90) |
            (monthly_df["end_lng"].abs() > 180)
        ).sum()

        print(
            f"{filename}: "
            f"{invalid_start_coords} invalid start coordinates, "
            f"{invalid_end_coords} invalid end coordinates"
        )
```

```python
# Convert timestamps to datetime for each monthly dataset
```

```python
# Load and standardise all monthly datasets
monthly_dataframes = []

for filename in sorted(os.listdir(csv_path)):
  if filename.endswith(".csv"):
    monthly_df = pd.read_csv(f"{csv_path}/{filename}")

    monthly_df["started_at"] = pd.to_datetime(monthly_df["started_at"])
    monthly_df["ended_at"] = pd.to_datetime(monthly_df["ended_at"])

    monthly_dataframes.append(monthly_df)
```

```python
# Verify timestamp data types after conversion
monthly_dataframes[0][["started_at", "ended_at"]].dtypes
```

```python
# Combine all monthly datasets into one DataFrame
df = pd.concat(monthly_dataframes, ignore_index=True)
```

```python
# Check the combined dataset
df.shape
```

```python
# Inspect the combined dataset
df.info()
```

```python
# Check missing values in the combined dataset
df.isnull().sum()
```

```python
# Check for duplicate ride IDs across the combined dataset
duplicate_ride_ids = df["ride_id"].duplicated().sum()

duplicate_ride_ids
```

```python
# Check categorical values in the combined dataset
print("rideable_type:")
print(df["rideable_type"].unique())

print("\nmember_casual:")
print(df["member_casual"].unique())
```

```python
# Validate geographic coordinate ranges
invalid_start_coords = (
    (df["start_lat"].abs() > 90) |
    (df["start_lng"].abs() > 180)
).sum()

invalid_end_coords = (
    (df["end_lat"].abs() > 90) |
    (df["end_lng"].abs() > 180)
).sum()

print("Invalid start coordinates:", invalid_start_coords)
print("Invalid end coordinates:", invalid_end_coords)
```

```python
# Check the date range covered by the combined dataset
print("Start date:", df["started_at"].min())
print("End date:", df["started_at"].max())
```

```python
# Calculate ride duration in minutes
df["ride_duration_minutes"] = (
    df["ended_at"] - df["started_at"]
).dt.total_seconds() / 60

df["ride_duration_minutes"].head()
```

```python
# Identify rides with negative calculated durations
negative_duration_count = (
    df["ride_duration_minutes"] < 0
).sum()

negative_duration_count
```

```python
# Flag rides with negative calculated durations
df["invalid_duration"] = df["ride_duration_minutes"] < 0

df["invalid_duration"].value_counts()
```

```python
# Summarise ride duration
df["ride_duration_minutes"].describe()
```

```python
# Count rides longer than one hour
long_ride_count = (
    df["ride_duration_minutes"] > 60
).sum()

long_ride_count
```

```python
# Count rides longer than 24 hours
very_long_ride_count = (
    df["ride_duration_minutes"] > 24 * 60
).sum()

very_long_ride_count
```

```python
# Inspect the longest rides
df["ride_duration_minutes"].nlargest(10)
```

```python
# Inspect the longest ride
df.loc[
    df["ride_duration_minutes"].idxmax(),
    ["ride_id", "started_at", "ended_at", "ride_duration_minutes", "member_casual"]
]
```

```python
# Create an analysis-ready duration field
df["ride_duration_clean"] = df["ride_duration_minutes"].where(
    ~df["invalid_duration"]
)

df["ride_duration_clean"].describe()
```

```python
# Extract the month of each ride
df["ride_month"] = df["started_at"].dt.to_period("M")

df["ride_month"].head()
```

```python
# Extract the day of the week for each ride
df["ride_day_of_week"] = df["started_at"].dt.day_name()

df["ride_day_of_week"].head()
```

```python
# Extract the starting hour for each ride
df["ride_start_hour"] = df["started_at"].dt.hour

df["ride_start_hour"].head()
```

```python
# Save the cleaned dataset with derived features
output_path = "/content/drive/MyDrive/cyclistic-data-cleaning-practice/data/cyclistic_cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully.")
```
